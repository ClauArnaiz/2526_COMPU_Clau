## Fundamento teórico

En mecánica cuántica, el estado de una partícula en una dimensión queda completamente descrito por una función de onda $\phi(x,t)$, cuya evolución temporal viene determinada por la ecuación de Schrödinger dependiente del tiempo:

$$
i \frac{\partial \phi(x,t)}{\partial t} = \left[-\frac{\partial^2}{\partial x^2} + V(x)\right]\phi(x,t)
$$

donde se han utilizado unidades reescaladas en las que $\hbar = 1$ y $m = 1/2$. La función de onda contiene toda la información del sistema, y su módulo al cuadrado define la densidad de probabilidad:

$$
P(x,t) = |\phi(x,t)|^2
$$

La interpretación probabilística exige que la función de onda esté normalizada en todo instante de tiempo:

$$
\int |\phi(x,t)|^2 dx = 1
$$

En esta práctica se considera un potencial de oscilador armónico cuántico de la forma:

$$
V(x) = \frac{\omega^2}{4}(x - 0.5)^2
$$

centrado en $x = 0.5$. Este sistema es especialmente relevante, ya que sus soluciones estacionarias (autofunciones del Hamiltoniano) son conocidas analíticamente y presentan energías cuantizadas:

$$
E_n = \omega \left(n + \frac{1}{2}\right)
$$

Estas autofunciones evolucionan en el tiempo mediante un factor de fase complejo, lo que implica que la densidad de probabilidad $|\phi(x,t)|^2$ es independiente del tiempo para estos estados.

Además de las autofunciones, también se estudian funciones de onda de tipo gaussiano, que no son estados estacionarios. En este caso, la distribución de probabilidad sí evoluciona en el tiempo, mostrando un comportamiento dinámico dentro del potencial.

Para caracterizar el sistema, se analizan distintos valores esperados asociados a observables físicos. 
- El valor medio de la posición viene dado por:

$$
\langle x(t) \rangle = \int x |\phi(x,t)|^2 dx
$$

- El valor medio del momento se define como:

$$
\langle p(t) \rangle = \int \phi^*(x,t)\left(-i \frac{\partial}{\partial x}\right)\phi(x,t)\,dx
$$

- Por su parte, el valor medio de la energía se obtiene a partir del Hamiltoniano:

$$
\langle H(t) \rangle = \int \phi^*(x,t)\left[-\frac{\partial^2}{\partial x^2} + V(x)\right]\phi(x,t)\,dx
$$

- Otra magnitud de interés es el producto de incertidumbres en posición y momento:

$$
\Delta x \, \Delta p
$$

donde:

$$
\Delta x = \sqrt{\langle x^2 \rangle - \langle x \rangle^2}, \quad
\Delta p = \sqrt{\langle p^2 \rangle - \langle p \rangle^2}
$$

Este producto debe cumplir el principio de incertidumbre de Heisenberg:

$$
\Delta x \, \Delta p \geq \frac{1}{2}
$$

Finalmente, se compara el comportamiento cuántico con el de un oscilador armónico clásico, donde la partícula presenta un movimiento periódico. Esta comparación permite analizar el principio de correspondencia, según el cual el comportamiento cuántico tiende al clásico en ciertos límites.

El sistema se discretiza tomando:

$$
x_j = j\Delta x, \qquad t_n = n\Delta t
$$

donde $j = 0,1,\dots,S$ y $n = 0,1,2,\dots$. En esta práctica se toma una caja de longitud $L=1$ con $S=1000$, por lo que:

$$
\Delta x = \frac{L}{S} = 0.001
$$

Además, se utiliza un paso temporal:

$$
\Delta t = 0.0001
$$

y se simulan $5000$ pasos temporales, de forma que el tiempo total estudiado es:

$$
t_{\text{max}} = 5000 \Delta t = 0.5
$$

Las condiciones de contorno consideradas son las correspondientes a paredes infinitas en los extremos de la caja:

$$
\phi(0,t)=0, \qquad \phi(1,t)=0
$$

El Hamiltoniano discretizado se escribe como:

$$
H_D \phi_j =
-\frac{\phi_{j+1}-2\phi_j+\phi_{j-1}}{(\Delta x)^2}
+ V_j \phi_j
$$

donde el potencial utilizado en este voluntario es el del oscilador armónico cuántico:

$$
V_j = \frac{\omega^2}{4}(x_j-0.5)^2
$$

con $\omega = 200$.

Para la evolución temporal se emplea la aproximación de Cayley:

$$
\phi^{n+1} =
\frac{1 - i\Delta t H_D/2}
{1 + i\Delta t H_D/2}
\phi^n
$$

Este método es especialmente adecuado porque conserva la norma de la función de onda, lo cual es necesario para mantener la interpretación probabilística del sistema.

En la práctica, la expresión anterior se transforma en un sistema tridiagonal que se resuelve en cada paso temporal. Una vez obtenida la solución auxiliar, la función de onda se actualiza iterativamente hasta alcanzar el tiempo final.

### Definición del potencial armónico

En este bloque se define el potencial al que está sometida la partícula. En concreto, se considera un potencial de oscilador armónico cuántico de la forma:

$$
V(x) = \frac{\omega^2}{4}(x - x_c)^2
$$

donde $\omega$ es la frecuencia angular del oscilador y $x_c$ representa la posición del mínimo del potencial, que en este caso se sitúa en el centro de la caja.

Este tipo de potencial describe una fuerza restauradora proporcional al desplazamiento respecto al equilibrio, lo que da lugar a un comportamiento oscilatorio de la partícula. Se trata de uno de los sistemas más relevantes en mecánica cuántica, ya que permite estudiar propiedades fundamentales como la cuantización de la energía.

La implementación se realiza mediante una función que evalúa el potencial en cada punto de la malla espacial, generando así un array que se utilizará posteriormente en la evolución temporal de la función de onda.

```python
# ============================================================
# 2. POTENCIAL ARMÓNICO
# ============================================================

def potencial_armonico(x, omega=200, xc=0.5):
    return (omega**2 / 4) * (x - xc)**2

V = potencial_armonico(x, omega, xc)
```

###  Funciones auxiliares

En este bloque se definen diversas funciones auxiliares necesarias para el tratamiento numérico de la función de onda y el cálculo de las magnitudes físicas del sistema.

En primer lugar, se implementa una función de normalización de la función de onda. Dado que la interpretación probabilística exige que la norma sea igual a la unidad, se calcula:

$$
\int |\phi(x,t)|^2 dx \approx \sum |\phi_j|^2 \Delta x
$$

y se reescala la función de onda para asegurar dicha condición.

A continuación, se definen funciones para calcular la derivada primera y segunda mediante diferencias finitas. La derivada primera se aproxima utilizando una fórmula centrada en los puntos interiores y diferencias hacia adelante y atrás en los extremos. Por su parte, la derivada segunda se calcula mediante la expresión:

$$
\frac{\partial^2 \phi}{\partial x^2} \approx \frac{\phi_{j+1} - 2\phi_j + \phi_{j-1}}{\Delta x^2}
$$

Estas derivadas son fundamentales para evaluar el operador momento y el Hamiltoniano del sistema.

Finalmente, se define una función que calcula los observables físicos más relevantes. En particular, se obtienen:

- La norma de la función de onda.
- El valor medio de la posición $\langle x \rangle$.
- El valor medio del momento $\langle p \rangle$.
- El valor medio de la energía $\langle H \rangle$.
- Las incertidumbres en posición y momento, $\Delta x$ y $\Delta p$.

Estos valores se calculan discretizando las integrales correspondientes en la malla espacial. El operador momento se implementa como $-i \frac{\partial}{\partial x}$, mientras que la energía se obtiene aplicando el Hamiltoniano al estado.

El producto de incertidumbres $\Delta x \Delta p$ se calcula para verificar el cumplimiento del principio de incertidumbre de Heisenberg.

```python
# ============================================================
# 3. FUNCIONES AUXILIARES
# ============================================================
# Comprobamos la noramalización
def normalizar(phi):
    norma = np.sqrt(np.sum(np.abs(phi)**2) * dx)
    return phi / norma

# Definimos la primera derivada
def derivada_primera(phi):
    dphi = np.zeros_like(phi, dtype=complex)
    dphi[1:-1] = (phi[2:] - phi[:-2]) / (2 * dx)
    dphi[0] = (phi[1] - phi[0]) / dx
    dphi[-1] = (phi[-1] - phi[-2]) / dx
    return dphi

# Definimos la segunda derivada
def derivada_segunda(phi):
    d2phi = np.zeros_like(phi, dtype=complex)
    d2phi[1:-1] = (phi[2:] - 2 * phi[1:-1] + phi[:-2]) / dx**2
    return d2phi

# Definimos la función que calcula los observables físicos más relevantes
def observables(phi, V):
    prob = np.abs(phi)**2

    norma = np.sum(prob) * dx

    x_medio = np.sum(x * prob) * dx
    x2_medio = np.sum(x**2 * prob) * dx

    dphi = derivada_primera(phi)
    d2phi = derivada_segunda(phi)

    p_medio = np.sum(np.conj(phi) * (-1j * dphi)) * dx
    p2_medio = np.sum(np.conj(phi) * (-d2phi)) * dx

    Hphi = -d2phi + V * phi
    energia = np.sum(np.conj(phi) * Hphi) * dx

    delta_x = np.sqrt(np.real(x2_medio - x_medio**2))
    delta_p = np.sqrt(np.real(p2_medio - p_medio**2))

    return {
        "norma": np.real(norma),
        "x": np.real(x_medio),
        "p": np.real(p_medio),
        "energia": np.real(energia),
        "delta_x": delta_x,
        "delta_p": delta_p,
        "incertidumbre": delta_x * delta_p
    }
```

###  Autofunciones del oscilador armónico

En este bloque se implementan las autofunciones del Hamiltoniano del oscilador armónico cuántico, que constituyen los estados estacionarios del sistema.

Estas funciones vienen dadas, en términos generales, por una combinación de un factor gaussiano y los polinomios de Hermite. Por este motivo, en primer lugar se define una función que permite calcular dichos polinomios mediante una relación de recurrencia:

$$
H_{n}(z) = 2z\,H_{n-1}(z) - 2(n-1)\,H_{n-2}(z)
$$

con condiciones iniciales $H_0 = 1$ y $H_1 = 2z$. Esta implementación permite obtener de forma eficiente polinomios de orden arbitrario.

A continuación, se construyen las autofunciones del oscilador armónico utilizando la expresión:

$$
\phi_n(x) \propto H_n(z)\, e^{-z^2/2}
$$

donde se ha definido la variable adimensional:

$$
z = \sqrt{\alpha}(x - x_c), \quad \alpha = \frac{\omega}{2}
$$

El prefactor incluido en la implementación asegura la correcta normalización de la función de onda. Además, se imponen condiciones de contorno nulas en los extremos de la caja, de acuerdo con el modelo considerado.

Finalmente, se define una función que devuelve la energía teórica de cada nivel cuántico, dada por:

$$
E_n = \omega \left(n + \frac{1}{2}\right)
$$

Estas energías permiten comparar los resultados numéricos obtenidos con las predicciones analíticas, verificando así la validez del método.

```python
# ============================================================
# 4. AUTOFUNCIONES DEL OSCILADOR ARMÓNICO
# ============================================================
# Definimos la función que permite calcular los polinomios de Hermite mediante una relación de recurrencia
def hermite(n, z):
    if n == 0:
        return np.ones_like(z)
    if n == 1:
        return 2 * z

    H0 = np.ones_like(z)
    H1 = 2 * z

    for k in range(2, n + 1):
        H2 = 2 * z * H1 - 2 * (k - 1) * H0
        H0, H1 = H1, H2

    return H1

# Definimos las autofunciones del oscilador
def autofuncion_oscilador(n, x, omega=200, xc=0.5):
    alpha = omega / 2
    z = np.sqrt(alpha) * (x - xc)

    prefactor = (alpha / np.pi)**0.25 / math.sqrt((2**n) * math.factorial(n))
    phi = prefactor * hermite(n, z) * np.exp(-z**2 / 2)

    phi[0] = 0
    phi[-1] = 0

    return normalizar(phi.astype(complex))

# Definimos la función que devuelve la energía teórica de cada nivel cuántico
def energia_teorica(n, omega=200):
    return omega * (n + 0.5)
```

### Método de Cayley

En este bloque se implementa el método de evolución temporal utilizado para resolver numéricamente la ecuación de Schrödinger. Se trata del mismo esquema desarrollado en la fase obligatoria, basado en la aproximación de Cayley:

$$
\phi^{n+1} =
\frac{1 - i\Delta t H_D/2}
{1 + i\Delta t H_D/2}
\phi^n
$$

La ventaja principal de este método es que el operador de evolución es unitario, por lo que conserva la norma de la función de onda durante la simulación. Esto es fundamental, ya que la norma representa la probabilidad total de encontrar la partícula en algún punto del espacio.

Para aplicar este método, la ecuación se transforma en un sistema tridiagonal que debe resolverse en cada paso temporal. La función `preparar_cayley` calcula los coeficientes necesarios para resolver dicho sistema. En particular, se define:

$$
w = \frac{\Delta t}{\Delta x^2}
$$

y se reescala el potencial como:

$$
\tilde{V}_j = \Delta x^2 V_j
$$

A partir de estos valores se construye el coeficiente central del sistema tridiagonal:

$$
A^0_j = -2 + \frac{2i}{w} - \tilde{V}_j
$$

Además, se calculan los coeficientes $\alpha_j$, que solo dependen del potencial y de los parámetros de discretización, por lo que basta con obtenerlos una vez al inicio.

La función `evolucionar` realiza un paso temporal completo. Primero calcula los coeficientes $\beta_j$ mediante una recurrencia hacia atrás, después obtiene la función auxiliar $q_j$ mediante una recurrencia hacia delante y finalmente actualiza la función de onda como:

$$
\phi^{n+1}_j = q_j - \phi^n_j
$$

También se imponen de nuevo las condiciones de contorno:

$$
\phi(0,t)=0, \qquad \phi(1,t)=0
$$

De esta forma, el código permite avanzar la función de onda paso a paso manteniendo la estabilidad numérica y la conservación de la norma.

```python
# ============================================================
# 5. MÉTODO DE CAYLEY
# ============================================================
# Definimos el método de evolución temporal
def preparar_cayley(V):
    w = dt / dx**2
    V_tilde = dx**2 * V

    # Construcción del coeficiente central del sistema tridiagonal
    A0 = -2 + 2j / w - V_tilde

    # Cálculo de los coeficientes alpha
    alpha = np.zeros(S + 1, dtype=complex)
    alpha[S - 1] = 0

    for jj in range(S - 1, 0, -1):
        gamma = 1 / (A0[jj] + alpha[jj])
        alpha[jj - 1] = -gamma

    return alpha, A0, w

alpha, A0, w = preparar_cayley(V)

# Definimos la función evolucionar para realizar un paso temporal completo
def evolucionar(phi, alpha, A0, w):
    beta = np.zeros(S + 1, dtype=complex)
    beta[S - 1] = 0

    for jj in range(S - 1, 0, -1):
        b = 4j * phi[jj] / w
        gamma = 1 / (A0[jj] + alpha[jj])
        beta[jj - 1] = gamma * (b - beta[jj])

    q = np.zeros(S + 1, dtype=complex)
    q[0] = 0
    q[-1] = 0

    for jj in range(0, S):
        q[jj + 1] = alpha[jj] * q[jj] + beta[jj]

    # Actualización de la función de onda
    phi_new = q - phi
    # Condiciones de contorno
    phi_new[0] = 0
    phi_new[-1] = 0

    return phi_new
```

### Simulación general

En este bloque se define una función general encargada de realizar la evolución temporal completa de la función de onda.

La función `simular` recibe como argumentos la función de onda inicial, el potencial considerado y un nombre identificativo para la simulación. En primer lugar, se calculan los coeficientes necesarios para aplicar el método de Cayley mediante la función `preparar_cayley`.

A continuación, se inicializan distintas listas donde se almacenarán los resultados de la simulación. En concreto, se guardan la función de onda, el tiempo correspondiente y los observables físicos principales: norma, posición media, momento medio, energía media e incertidumbre.

Durante la evolución temporal, el programa avanza la función de onda aplicando sucesivamente la función `evolucionar`. No obstante, para reducir el coste de memoria, no se guardan todos los pasos temporales, sino únicamente uno de cada cierto número de iteraciones, determinado por la variable `guardar_cada`.

En cada instante guardado se calculan los observables mediante la función `observables`, lo que permite estudiar posteriormente su evolución temporal y comprobar propiedades físicas como la conservación de la norma y de la energía.

Finalmente, la función devuelve todos los datos organizados en un diccionario, facilitando su posterior representación gráfica y análisis.

```python
# ============================================================
# 6. SIMULACIÓN GENERAL
# ============================================================
def simular(phi0, V, nombre="simulacion"):
    alpha, A0, w = preparar_cayley(V)

    phi = phi0.copy()

    phis = []
    tiempos = []

    normas = []
    posiciones = []
    momentos = []
    energias = []
    incertidumbres = []

    for n in range(nsteps + 1):
        if n % guardar_cada == 0:
            phis.append(phi.copy())
            tiempos.append(n * dt)

            obs = observables(phi, V)
            normas.append(obs["norma"])
            posiciones.append(obs["x"])
            momentos.append(obs["p"])
            energias.append(obs["energia"])
            incertidumbres.append(obs["incertidumbre"])

        if n < nsteps:
            phi = evolucionar(phi, alpha, A0, w)

    return {
        "nombre": nombre,
        "phis": np.array(phis),
        "tiempos": np.array(tiempos),
        "normas": np.array(normas),
        "posiciones": np.array(posiciones),
        "momentos": np.array(momentos),
        "energias": np.array(energias),
        "incertidumbres": np.array(incertidumbres)
    }
```

### Comparación con el oscilador clásico

En este bloque se implementa la solución correspondiente al oscilador armónico clásico, con el objetivo de comparar su comportamiento con el sistema cuántico.

A partir de los valores iniciales de posición y momento obtenidos en la simulación cuántica, se construyen las trayectorias clásicas de la partícula. En el caso del oscilador armónico, estas vienen dadas por soluciones sinusoidales:

$$
x(t) = x_c + (x_0 - x_c)\cos(\omega t) + \frac{2p_0}{\omega}\sin(\omega t)
$$

$$
p(t) = p_0 \cos(\omega t) - \frac{\omega}{2}(x_0 - x_c)\sin(\omega t)
$$

donde $x_0$ y $p_0$ representan las condiciones iniciales del sistema.

Además, la energía clásica se mantiene constante a lo largo del tiempo, por lo que se toma igual al valor inicial obtenido en la simulación.

Esta comparación permite analizar el principio de correspondencia, según el cual el comportamiento del sistema cuántico tiende al clásico en determinados regímenes, especialmente para valores grandes del número cuántico o en estados tipo gaussiano bien localizados.

```python
# ============================================================
# 9. COMPARACIÓN CON OSCILADOR CLÁSICO
# ============================================================

def teorico_clasico(resultado):
    t = resultado["tiempos"]

    x0 = resultado["posiciones"][0]
    p0 = resultado["momentos"][0]
    E0 = resultado["energias"][0]

    x_cl = xc + (x0 - xc) * np.cos(omega * t) + (2 * p0 / omega) * np.sin(omega * t)
    p_cl = p0 * np.cos(omega * t) - (omega / 2) * (x0 - xc) * np.sin(omega * t)
    E_cl = np.ones_like(t) * E0

    return {
        "x": x_cl,
        "p": p_cl,
        "E": E_cl
    }
```

$$

### Principio de correspondencia

En este último bloque se estudia el principio de correspondencia de Bohr, según el cual el comportamiento de un sistema cuántico debe aproximarse al comportamiento clásico en el límite de números cuánticos grandes.

Para ello, se calcula la autofunción del oscilador armónico correspondiente a un número cuántico alto, en este caso $n=20$, y se representa su densidad de probabilidad:

$$
P_n(x)=|\phi_n(x)|^2
$$

A continuación, se construye una distribución clásica equivalente. En el oscilador armónico clásico, la partícula se mueve entre dos puntos de retorno, donde su velocidad se anula. Estos puntos se obtienen imponiendo:

$$
E = \frac{\omega^2}{4}(x-x_c)^2
$$

de donde:

$$
|x-x_c| = \frac{2\sqrt{E}}{\omega}
$$

La probabilidad clásica es mayor cerca de los puntos de retorno, ya que la partícula clásica pasa más tiempo en esas regiones al moverse más lentamente. Por ello, la distribución clásica es proporcional a:

$$
P_{\text{cl}}(x) \propto \frac{1}{v(x)}
$$

En el código se calcula esta distribución y se normaliza para poder compararla con la probabilidad cuántica. La comparación gráfica permite observar cómo, para valores grandes de $n$, la envolvente de la distribución cuántica se aproxima al comportamiento clásico, aunque conserva las oscilaciones propias de la descripción cuántica.

```python
# ============================================================
# 13. PRINCIPIO DE CORRESPONDENCIA: n GRANDE
# ============================================================

# IMPORTANTE:
# Si antes daba error con n=20, asegúrate de que la función
# autofuncion_oscilador usa math.sqrt en el prefactor, no np.sqrt.

n_grande = 20

# Autofunción del oscilador armónico para n grande
phi_n20 = autofuncion_oscilador(n_grande, x, omega=200, xc=0.5)
prob_n20 = np.abs(phi_n20)**2

# Energía teórica del nivel n=20
E20 = energia_teorica(n_grande, omega=200)

# Puntos de retorno clásicos
# Para V(x) = omega^2/4 (x - xc)^2
# E = omega^2/4 (x - xc)^2
# |x - xc| = 2 sqrt(E) / omega
x_min = xc - 2 * np.sqrt(E20) / 200
x_max = xc + 2 * np.sqrt(E20) / 200

# Probabilidad clásica proporcional a 1/v(x)
prob_cl = np.zeros_like(x)

eps = 1e-6
zona = (x > x_min + eps) & (x < x_max - eps)

prob_cl[zona] = 1 / np.sqrt((x[zona] - x_min) * (x_max - x[zona]))

# Normalizar la probabilidad clásica
prob_cl = prob_cl / (np.sum(prob_cl) * dx)

# Gráfica comparativa
plt.figure(figsize=(8, 4))
plt.plot(x, prob_n20, label="Probabilidad cuántica n=20")
plt.plot(x, prob_cl, "--", label="Probabilidad clásica reescalada")
plt.xlabel("x")
plt.ylabel("Probabilidad")
plt.title("Principio de correspondencia para n = 20")
plt.legend()
plt.grid()
plt.show()
```

### Principio de correspondencia

En este apartado se analiza el comportamiento del oscilador armónico cuántico en el límite de números cuánticos elevados con el objetivo de verificar experimentalmente el Principio de Correspondencia de Bohr. Para ello, se ejecutaba el algoritmo utilizando la autofunción del nivel $n = 20$  y se contrastaba su densidad de probabilidad cuántica, $P_{20}(x) = |\phi_{20}(x)|^2$, con la distribución de probabilidad clásica equivalente deducida a partir de la mecánica newtoniana. 

En un oscilador armónico clásico, la probabilidad de encontrar la partícula en un diferencial de posición $dx$ es inversamente proporcional a su velocidad instantánea, $P_{\text{cl}}(x) \propto 1/v(x)$, debido a que el sistema pasa una mayor fracción de tiempo en aquellas regiones donde se desplaza más lentamente.

A partir de la conservación de la energía clásica, $E = \frac{p^2}{2m} + V(x)$, y considerando la masa reescalada del sistema $m = 1/2$, la velocidad se expresa como:

$$v(x) = 2\sqrt{E - V(x)}$$

Sustituyendo el potencial armónico cuadrático $V(x) = \frac{\omega^2}{4}(x-x_c)^2$, se determinan analíticamente los puntos de retorno clásicos ($x_{\text{min}}$ y $x_{\text{max}}$) imponiendo $v(x) = 0$:
  $$|x - x_c| = \frac{2\sqrt{E}}{\omega}$$
  
 La densidad de probabilidad clásica normalizada a la unidad toma la forma asintótica que implementa el código:
 
 $$P_{\text{cl}}(x) = \frac{1}{\pi \sqrt{(x - x_{\text{min}})(x_{\text{max}} - x)}}$$
 
 
Al observar la gráfica se observa el comportamiento ondulatoria residual. Ya que, la distribución cuántica $P_{20}(x)$ exhibe un patrón altamente oscilatorio con exactamente $20$ nodos espaciales en los que la probabilidad de encontrar a la partícula se anula por completo. Este fenómeno es puramente cuántico y carece de contrapartida clásica.

La envolvente del paquete de ondas cuántico se acopla a la curva teórica de la probabilidad clásica. Ambas funciones divergen simétricamente hacia el infinito en las proximidades de $x_{\text{min}}$ y $x_{\text{max}}$, reflejando que tanto el oscilador clásico como el cuántico altamente excitado concentran su probabilidad en los extremos del movimiento, donde la partícula frena para invertir el sentido de la marcha.

En los bordes externos a los puntos de retorno ($x < x_{\text{min}}$ ó $x > x_{\text{max}}$), la probabilidad clásica cae estrictamente a cero, es decir, estamos ante un región mecánicamente prohibida. Sin embargo, la función cuántica muestra una pequeña cola de penetración exponencial decreciente. Aunque para $n=20$, este efecto de confinamiento cuántico ya es despreciable, confinando casi la totalidad de la norma en la zona clásica.

En resumen, los resultados simulados constituyen una verificación numérica directa del postulado de Bohr. Aunque el formalismo cuántico mantiene sus propiedades intrínsecas, el comportamiento promediado a gran escala de la densidad de probabilidad tiende macroscópicamente al régimen clásico conforme el número cuántico principal se incrementa ($n \ge 20$). 

# Coste computacional

```python
# ============================================================
# ESTUDIO DEL COSTE COMPUTACIONAL
# ============================================================

import time
import numpy as np
import matplotlib.pyplot as plt

def medir_coste_computacional(lista_S, n_pasos_test=500):
    """
    Mide el tiempo de ejecución del método de Cayley para distintos
    tamaños de malla espacial S, manteniendo fijo el número de pasos
    temporales de prueba.
    """

    tiempos = []

    for S_test in lista_S:
        print(f"Calculando tiempo para S = {S_test}...")

        # Parámetros locales
        L_test = 1.0
        dx_test = L_test / S_test
        x_test = np.linspace(0, L_test, S_test + 1)

        dt_test = 1e-4
        omega_test = 200
        xc_test = 0.5

        # Potencial armónico
        V_test = (omega_test**2 / 4) * (x_test - xc_test)**2

        # Gaussiana inicial
        x0_test = 0.3
        sigma_test = 1 / 16
        phi = np.exp(-(x_test - x0_test)**2 / (2 * sigma_test**2)).astype(complex)
        phi[0] = 0
        phi[-1] = 0

        # Normalización
        norma = np.sqrt(np.sum(np.abs(phi)**2) * dx_test)
        phi = phi / norma

        # ====================================================
        # Preparación local del método de Cayley
        # ====================================================

        w_test = dt_test / dx_test**2
        V_tilde = dx_test**2 * V_test

        A0_test = -2 + 2j / w_test - V_tilde

        alpha_test = np.zeros(S_test + 1, dtype=complex)
        alpha_test[S_test - 1] = 0

        for jj in range(S_test - 1, 0, -1):
            gamma = 1 / (A0_test[jj] + alpha_test[jj])
            alpha_test[jj - 1] = -gamma

        # ====================================================
        # Evolución temporal
        # ====================================================

        inicio = time.perf_counter()

        for _ in range(n_pasos_test):

            beta = np.zeros(S_test + 1, dtype=complex)
            beta[S_test - 1] = 0

            for jj in range(S_test - 1, 0, -1):
                b = 4j * phi[jj] / w_test
                gamma = 1 / (A0_test[jj] + alpha_test[jj])
                beta[jj - 1] = gamma * (b - beta[jj])

            q = np.zeros(S_test + 1, dtype=complex)
            q[0] = 0
            q[-1] = 0

            for jj in range(0, S_test):
                q[jj + 1] = alpha_test[jj] * q[jj] + beta[jj]

            phi = q - phi
            phi[0] = 0
            phi[-1] = 0

        fin = time.perf_counter()

        tiempos.append(fin - inicio)

    return np.array(tiempos)


# Valores de S que se van a probar
lista_S = np.array([200, 400, 600, 800, 1000])

# Medimos el tiempo
tiempos_S = medir_coste_computacional(lista_S, n_pasos_test=500)

# Ajuste lineal para ver la tendencia
coef = np.polyfit(lista_S, tiempos_S, 1)
ajuste = np.polyval(coef, lista_S)

# Gráfica
plt.figure(figsize=(7, 4))
plt.plot(lista_S, tiempos_S, "o-", label="Tiempo medido")
plt.plot(lista_S, ajuste, "--", label="Ajuste lineal")
plt.xlabel("Número de puntos espaciales S")
plt.ylabel("Tiempo de ejecución (s)")
plt.title("Coste computacional en función del tamaño de la malla")
plt.grid()
plt.legend()
plt.show()

# Mostramos los tiempos por pantalla
print("\nTiempos medidos:")
for S_val, t_val in zip(lista_S, tiempos_S):
    print(f"S = {S_val:4d} -> tiempo = {t_val:.4f} s")

print("\nPendiente del ajuste lineal:", coef[0])
print("Ordenada en el origen:", coef[1])
```

Para analizar el coste computacional del algoritmo se ha medido el tiempo de ejecución al variar el número de puntos espaciales $S$, manteniendo fijo el número de pasos temporales utilizados en la prueba. Esta comparación permite estudiar cómo aumenta el tiempo de cálculo al refinar la malla espacial.

En cada paso temporal, el algoritmo resuelve un sistema lineal tridiagonal asociado al método de Cayley. La resolución de un sistema tridiagonal tiene un coste aproximadamente lineal con el tamaño del sistema, es decir,

$$
\text{coste por paso} \sim S.
$$

Como la evolución completa realiza $N_t$ pasos temporales, el coste total escala aproximadamente como

$$
\text{coste total} \sim S \, N_t.
$$

Por tanto, si el número de pasos temporales se mantiene fijo, se espera que el tiempo de ejecución aumente de forma aproximadamente lineal con $S$.

La gráfica obtenida confirma esta tendencia: al aumentar el número de puntos espaciales, el tiempo de ejecución crece de manera casi lineal. Esto indica que el uso de un algoritmo tridiagonal es eficiente, ya que evita la inversión de matrices completas y permite trabajar con mallas finas, como $S=1000$, sin que el coste computacional sea excesivo.

Por tanto, los resultados no son exactamente iguales a los valores analíticos, aunque deben aproximarse a ellos si la malla espacial y el paso temporal son suficientemente pequeños.

En este trabajo se ha utilizado una resolución espacial con $S=1000$ puntos y $\Delta x=0.001$ junto con un paso temporal pequeño, $\Delta t=10^{-4}$. La energía media también se mantiene constante en el tiempo para las autofunciones del Hamiltoniano. Además, los valores numéricos obtenidos para las energías de los primeros niveles coinciden muy bien con los valores teóricos

$$
\text{Error relativo} =

Los errores relativos obtenidos son muy pequeños, del orden de $10^{-3}\%$, lo que confirma que la discretización elegida es suficientemente precisa para reproducir correctamente los autovalores del oscilador armónico.

También se observan errores muy pequeños en los valores medios de la posición y del momento. Para las autofunciones, el valor medio de la posición debe permanecer en el centro del potencial,

Las pequeñas oscilaciones o desviaciones que aparecen en las gráficas no tienen significado físico relevante, sino que proceden de la discretización numérica, de los efectos de borde de la caja finita y de los errores de redondeo acumulados durante la evolución temporal.

En el caso de los paquetes gaussianos existen errores numéricos asociados a la discretización y a las diferencias físicas reales, ya que el paquete cuántico no es una partícula puntual clásica, sino una distribución espacial con anchura finita.

En cambio, si el paquete está centrado en $x_0=0.5$, su centro no se mueve, aunque la forma del paquete puede cambiar si su anchura no coincide con la del estado fundamental.

También se ha visto que al cambiar la frecuencia $\omega$ cambia la rapidez del movimiento. Si $\omega$ es mayor, el potencial es más rígido y el paquete oscila más rápido. Si $\omega$ es menor, el potencial es más suave y la oscilación es más lenta.

Por último, se ha comprobado la relación entre el comportamiento cuántico y el clásico. Los valores medios de posición y momento del paquete gaussiano siguen la trayectoria clásica esperada, y para un nivel excitado alto, como $n=20$, la probabilidad cuántica se parece cada vez más a la distribución clásica. Esto permite visualizar el principio de correspondencia de Bohr.
