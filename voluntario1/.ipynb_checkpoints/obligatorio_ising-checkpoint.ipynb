# Fundamento teórico

La Mecánica Estadística es la rama de la física que utiliza la teoría de la probabilidad y la estadística para explicar el comportamiento macroscópico (temperatura, presión, entropía) de sistemas físicos, basándose en el movimiento y las interacciones de sus componentes microscópicos individuales, como átomos o moléculas. Es decir, esta disciplina sirve como puente para relacionar la mecánica clásica/cuántica (microscópica) y la termodinámica (macroscópica).

En un sistema en equilibrio termodinámico (definido por una temperatura $T$, número de partículas $N$ y volumen $V$), la probabilidad de encontrar al sistema en un estado específico $X$ sigue la distribución de Boltzmann:

$$P_{N}(X)=Z^{-1}e^{-\beta H_{N}(X)}$$
donde:
- $H_{N}(X)$ es el Hamiltoniano que define la dinámica microscópica del sistema.
- $\beta = 1 / (k_B T)$, siendo $k_B$ la constante de Boltzmann.
- $Z$ es la función de partición, que actúa como factor de normalización.

A través de esta distribución, el valor esperado de cualquier variable observable $A$ (que coincide con el valor medido experimentalmente) se calcula como el promedio sobre todos los estados posibles:
$$\langle A\rangle=Z^{-1}\int dXe^{-\beta H_{N}(X)}A_{N}(X)$$

## El Modelo de Ising
Propuesto originalmente por Lenz en 1927 y estudiado por su estudiante Ising, este modelo busca explicar de forma sencilla el comportamiento de materiales que sufren cambios bruscos en sus propiedades magnéticas según la temperatura.

El modelo describe la transición entre un estado **paramagnético** a altas temperaturas (magnetización espontánea nula) y un estado **ferromagnético** a bajas temperaturas (magnetización espontánea no nula).

El sistema se define en una red cuadrada bidimensional de tamaño $N^2$. En cada nudo de la red existe una variable de espín $s(i,j)$ que solo puede tomar dos valores: $1$ (up spin) o $-1$ (down spin).



La energía de interacción de una configuración de espines $S$ en el modelo de Ising viene dada por la suma de las interacciones entre vecinos próximos:
$$E(S)=-\frac{1}{2}\sum_{i=1}^{N}\sum_{j=1}^{N}s(i,j)[s(i,j+1)+s(i,j-1)+s(i+1,j)+s(i-1,j)]$$
Para simular un sistema infinito y evitar efectos de borde, se aplican condiciones de contorno periódicas. Esto significa que los espines en los bordes de la red interactúan con los del borde opuesto (por ejemplo, $s(N+1,j)=s(1,j)$), tratando la red como si estuviera sobre la superficie de un toroide.

## Fenomenología del Modelo: La transición de fase
El principal interés del modelo de Ising se basa en su capacidad paradescribir de forma simplificada la transición de fase entre un estado paramagnético y uno ferromagnético. 

Este fenómeno surge de la competencia entre la energía de interacción de los espines y el desorden inducido por la temperatura:
- **A bajas temperaturas (Fase ferromagnética):** La energía de interacción predomina, favoreciendo que los espines vecinos se alineen entre sí. Esto da lugar a una magnetización espontánea no nula, donde el sistema presenta un orden macroscópico.
- **A altas temperaturas (Fase paramagnética):** El desorden térmico supera las fuerzas de alineamiento, provocando que las orientaciones de los espines sean aleatorias. En este régimen, la magnetización promedio es cero y el sistema pierde su ordenamiento magnético.

La identificación y caracterización de esta temperatura crítica, donde el sistema cambia abruptamente de comportamiento, constituye uno de los objetivos centrales de esta simulación computacional. Se conoce que la temperatura crítica es aproximadamente 2,269 (a menudo redondeada a 2,27) porque existe una solución analítica exacta para el modelo de Ising en dos dimensiones sobre una red cuadrada, la cual fue hallada por el físico Lars Onsager en 1944.

## Necesidad de Métodos Numéricos (Monte Carlo)
En la práctica, la resolución analítica de las integrales N-dimensionales necesarias para obtener los valores esperados de observables físicos resulta inviable para la mayoría de los sistemas complejos. Esto se debe a la enorme dimensionalidad del espacio de estados, lo que obliga a recurrir a estrategias de computación numérica para aproximar el comportamiento del sistema.

Aunque la Ley de los Grandes Números sugiere que es posible estimar valores medios a través de promedios sobre un conjunto de realizaciones independientes, la aplicación de técnicas de muestreo simple (como el método de rechazo) no es suficiente en sistemas de gran tamaño. Esto ocurre porque la distribución de probabilidad $P_N(X)$ se vuelve extremadamente estrecha y se concentra en torno a los valores de equilibrio térmico. En consecuencia, las configuraciones generadas de forma puramente aleatoria suelen poseer energías muy elevadas y una probabilidad de ocurrencia prácticamente nula, provocando que la inmensa mayoría de los intentos de muestreo sean rechazados. 

Para superar esta limitación, se emplean procesos más sofisticados basados en cadenas de Markov. Estas permiten explorar el espacio de configuraciones de manera dirigida, favoreciendo las regiones de alta probabilidad y evitando el muestreo de estados irrelevantes para la física del sistema.


# Método numérico: algoritmo de Metropolis
Para simular el comportamiento del modelo de Ising y estudiar sus propiedades termodinámicas, se emplea el método de Monte Carlo, concretamente el algoritmo de Metropolis. 

Esta técnica permite generar una secuencia de configuraciones del sistema distribuidas según el peso de Boltzmann sin la necesidad de explorar exhaustivamente el espacio de estados, cuya dimensionalidad hace que un cálculo directo sea computacionalmente inabordable.

El algoritmo de Metropolis se basa en la construcción de una cadena de Markov, donde cada nueva configuración del sistema se obtiene a partir de la anterior mediante pequeñas modificaciones locales. Este procedimiento garantiza que, tras un número suficiente de iteraciones (periodo de termalización), las configuraciones generadas sean representativas del equilibrio térmico.

Para asegurar la convergencia hacia la distribución de Boltzmann, las probabilidades de transición entre un estado inicial $X$ y un estado propuesto $X'$ deben cumplir la condición de balance detallado. Bajo este esquema, la probabilidad de aceptación $p$ para una transición viene dada por el criterio de Metropolis:
$$p = \min(1, e^{-\beta \Delta E})$$
donde $\Delta E = E(X') - E(X)$ 
representa el cambio de energía asociado a la transición y $\beta = 1/T$ (en unidades de $k_B = 1$).

## Aplicación al Modelo de Ising
En el contexto de este modelo, la dinámica del algoritmo se traduce en la inversión aleatoria de espines. La simulación sigue estos pasos operativos en cada iteración:
1. Se elige un espín $s_{i,j}$ de la red de forma aleatoria.
2. Se evalúa el cambio de energía $\Delta E$ que supondría invertir dicho espín. Para ello, se considera la interacción con sus cuatro vecinos más próximos y se aplican condiciones de contorno periódicas para simular un sistema infinito.
3. - Si $\Delta E \leq 0$, el cambio reduce o mantiene la energía del sistema, por lo que el giro se acepta siempre.
   - Si $\Delta E > 0$, el cambio se acepta únicamente con probabilidad $e^{-\Delta E/T}$. 
 
 Para implementar esto, se genera un número aleatorio uniforme $\xi \in [0, 1]$ y se acepta el giro solo si $\xi < e^{-\Delta E/T}$. Si el cambio es rechazado, el sistema permanece en la misma configuración para la siguiente iteración. Este mecanismo permite que el sistema explore el espacio de estados favoreciendo las regiones de mayor probabilidad (menor energía) sin quedar atrapado en mínimos locales gracias a la activación térmica.
   
Dado que en cada iteración individual solo se intenta modificar un único espín, es necesario definir una escala temporal adecuada para observar la evolución del sistema global. Se define el **Paso Monte Carlo (PMC)** como la unidad de tiempo de la simulación. Un PMC corresponde a $N^2$ intentos de actualización (donde $N^2$ es el número total de espines). De este modo, se asegura que, en promedio, cada espín de la red ha tenido la oportunidad de cambiar de signo una vez, permitiendo que la configuración global de la red evolucione de forma significativa.


# Código (explicado por partes)


El código se ha desarrollado en Python (más detallado en apartado (#Herramientas-de-programación)), utilizando la librería **numpy** para la manipulación eficiente de matrices y **matplotlib** para la generación de la interfaz gráfica y la animación en tiempo real. La implementación se divide en tres bloques principales:
## Inicialización y Parámetros

El sistema se define mediante una red bidimensional de tamaño $N \times N$. Se definen dos posibles estados iniciales:
- Estado ordenado: Todos los espines se inicializan con valor $+1$ (simulando un sistema a $T=0$).
- Estado desordenado: Los espines toman valores $+1$ o $-1$ de forma aleatoria con igual probabilidad (simulando un sistema a $T = \infty$).

```python
# ==========================================
# PARÁMETROS DEL SISTEMA
# ==========================================
N = 32          # Tamaño de la red (N x N)
T = 2.27        # Temperatura (puedes variarla en el intervalo [0, 5])
update_interval = 10  # Milisegundos entre fotogramas de la animación

# ==========================================
# PASO 0: INICIALIZACIÓN
# ==========================================
def inicializar_red(N, ordenada=False):
    """
    Genera la configuración inicial de espines.
    Puede ser ordenada (todos 1) o desordenada (aleatoria 1 o -1).
    """
    if ordenada:
        return np.ones((N, N))
    else:
        # Elegir aleatoriamente 1 o -1 con probabilidad 1/2
        return np.random.choice([-1, 1], size=(N, N))
```

## Implementación del Algoritmo de Metropolis

En esta parte se ejecutan $N^2$ intentos de cambio para completar un Paso Monte Carlo (PMC). Los aspectos técnicos clave son:
- <span style="color:#2E86C1">Condiciones de Contorno Periódicas:</span> Se implementan utilizando el operador módulo (% N). Esto permite que al evaluar los vecinos del espín $(n, m)$, si el índice sobrepasa el límite de la red, aparezca por el lado opuesto, eliminando así los efectos de borde.
- <span style="color:#2E86C1">Cálculo de la Diferencia de Energía ($\Delta E$): </span>Siguiendo el Hamiltoniano del modelo, la variación de energía al invertir un espín $s_{i,j}$ viene dada por:$$\Delta E = 2 \cdot s_{i,j} \sum s_{vecinos}$$ Esta fórmula es eficiente ya que solo requiere conocer el estado de los 4 vecinos más cercanos.
- <span style="color:#2E86C1">Decisión Estocástica:</span> Se implementa el criterio de aceptación comparando la probabilidad de Boltzmann con un número aleatorio uniforme generado mediante *np.random.rand()*.

```python
#  ==========================================
# PASO 1 A 4: ALGORITMO DE METROPOLIS (1 Paso Monte Carlo)
# ==========================================
def paso_monte_carlo(red, T):
    """
    Realiza 1 Paso Monte Carlo (PMC).
    En 1 PMC se realizan N^2 intentos de cambio de espín.
    """
    for _ in range(N * N):
        # Paso 1: Elegir un punto al azar (n, m) de la red
        n = np.random.randint(0, N)
        m = np.random.randint(0, N)
        
        esp_actual = red[n, m]
        
        # Condiciones de contorno periódicas: usamos módulo N
        vecino_arriba  = red[(n - 1) % N, m]
        vecino_abajo   = red[(n + 1) % N, m]
        vecino_izq     = red[n, (m - 1) % N]
        vecino_der     = red[n, (m + 1) % N]
        
        # Paso 2: Evaluar el cambio de energía (Delta E)
        suma_vecinos = vecino_arriba + vecino_abajo + vecino_izq + vecino_der
        delta_E = 2 * esp_actual * suma_vecinos
        
        # Evaluar probabilidad p = min(1, exp(-Delta_E / T))
        # Si delta_E <= 0, exp(-delta_E / T) es >= 1, por lo que p = 1 (se acepta siempre)
        if delta_E <= 0:
            red[n, m] = -esp_actual
        else:
            # Paso 3: Generar número aleatorio uniforme y comparar
            p = np.exp(-delta_E / T)
            if np.random.rand() < p:
                red[n, m] = -esp_actual # Cambiar el signo del espín

    # Paso 4: Ir a (1) (El bucle for se encarga de repetir esto N^2 veces)
    return red
```

## Visualización y Evolución Temporal

Esta última parte se usa para observar la dinámica del sistema (cómo se forman los dominios magnéticos). En cada fotograma de la animación, se llama a la función de Metropolis, actualizando la configuración de la red.
Se utiliza un mapa de colores (*coolwarm*) donde un color representa el espín $+1$ y otro el $-1$. Esto permite visualizar de forma intuitiva cómo, a temperaturas bajas, el sistema crea grandes conjuntos de espines alineados, mientras que a temperaturas altas el ruido térmico impide cualquier ordenamiento.

```python
# ==========================================
# CONFIGURACIÓN DE LA VISUALIZACIÓN
# ==========================================
# Inicializamos la red (empezamos con una desordenada por ejemplo)
red = inicializar_red(N, ordenada=False)

fig, ax = plt.subplots(figsize=(6, 6))
ax.set_title(f"Modelo de Ising - Evolución a T = {T}")
ax.axis('off')

# Mostramos la red inicial (1 se verá de un color, -1 de otro)
imagen = ax.imshow(red, cmap='coolwarm', interpolation='nearest', vmin=-1, vmax=1)

def actualizar(frame):
    """Función que actualiza el gráfico en cada fotograma de la animación."""
    global red
    red = paso_monte_carlo(red, T)
    imagen.set_data(red)
    return [imagen]

# Ejecutamos la animación
anim = animation.FuncAnimation(fig, actualizar, frames=200, interval=update_interval, blit=True)

plt.show()
```

## Tamaño de red N = 32
Usando el código explicado para el mismo tamaño de red (N=32) se estudiaron los tres posibles régimen del sistema:
1. Baja Temperatura (T=1)
2. Temperatura crítica (T=2.7)
3. Alta Temperatura (T=4.5)

A continuación, se muestra los resultados obtenidos:


Se observa que para bajas temperaturas que la red está prácticamente dominada por un solo color (azul), con apenas un par de píxeles rojos aislados. Esto se dede a que a esta temperatura, la energia térmica es muy baja y predomina la energía de interacción entre vecinos, por lo que el sistema tiende fuertemente al orden: los espines se han alineado masivamente en una misma dirección, logrando una magnetización espontánea alta. De este modo, confirmamos que exactamente estamos en la fase ferromagnética.

Para la temperatura crítica se pueden ver grandes cúmulos de ambos colores que van fluctuando. Las formas no son cuadradas ni homogéneas, sino que tienen un aspecto fractal. Esto se debe a que la temperatura $T = 2.27$ está extremadamente cerca de la temperatura crítica exacta calculada por Lars Onsager ($T_c \approx 2.269$). En este punto, la tendencia al orden (alinearse) y al desorden (ruido térmico) están perfectamente equilibradas. Es deci, las fluctuaciones ocurren a todas las escalas espaciales.

Y por último, a altas temperaturas, se observa que hay una mezcla caótica y homogénea de píxeles rojos y azules muy pqueños. Esto se explica ya que el desorden inducido por la temperatura supera por completo a las fuerzas de alineamiento magnético, por lo que no se forman dominios estables debido a que la agitación térmica voltea los espines constantemente. La magnetización neta prodemia del sistema es cero y nos encontramos por tanto es la fase paramagnética.

En las animaciones del [Apartado 5.4](##Animaciones-GIF), se puede observar cómo los bordes de estos dominios se suavizan y el sistema busca minimizar su energía interfacial, tendiendo a un estado monocromático si se deja el tiempo suficiente. Este comportamiento valida que el algoritmo minimiza la energía a bajas temperaturas.

Coexisten grandes islas con multitud de pequeñas fluctuaciones aisladas. Esta temperatura corresponde al punto crítico, donde tenemos el equilibrio perfecto entre la tendencia al orden (alinearse) y al desorden (agitación térmica). La agitación térmica es tan alta que "voltea" los espines constantemente, superando las fuerzas de alineamiento. La magnetización neta promedio en este estado es cero. 

## Comparación para N = 32 y N = 68
En el estudio del modelo de Ising, el tamaño de la red ($N \times N$) es un factor muy importante debido a lo que conocemos como Efectos de Tamaño Finito (Finite Size Effects). 

Aunque la fenomenología general es similar, el paso de una red de $N=32$ a $N=68$ introduce diferencias físicas y estadísticas fundamentales:

1. <u>Aproximación al Límite Termodinámico:</u> La física teórica define las transiciones de fase puras en el límite termodinámico, donde el número de partículas tiende a infinito ($N \to \infty$). En una simulación numérica, estamos limitados por la capacidad de cómputo. Al aumentar el tamaño a $N=68$, estamos reduciendo la discrepancia entre nuestro "modelo de juguete" y el comportamiento de un material real. Una red de $N=68$ contiene 4,624 espines, frente a los 1,024 de la red de $N=32$. Este incremento de casi el 450% en el número de agentes permite que las leyes estadísticas de los grandes números actúen con mayor rigor.

2. <u>Resolución de la Longitud de Correlación ($\xi$):</u> Cerca de la temperatura crítica ($T \approx 2.269$), la longitud de correlación $\xi$ (que mide el tamaño de los dominios de espines del mismo signo) tiende a divergir. 
- En $N=32$: Los dominios críticos crecen rápidamente hasta alcanzar el tamaño de la caja. Esto provoca que el sistema no pueda fluctuar de forma natural porque se topa con sus propias condiciones de contorno muy pronto.
- En $N=68$: El sistema tiene más espacio para desarrollar estructuras fractales complejas antes de verse limitado por el borde de la red, lo que permite capturar mejor la física de la transición.

3. <u>Nitidez de la Transición de Fase:</u> La transición de fase se manifiesta de forma más abrupta conforme $N$ aumenta:
- Magnetización ($m$): En $N=32$, la caída de la magnetización es más suave y redondeada. En $N=68$, la pendiente cerca de $T_c$ es más vertical, aproximándose a la discontinuidad teórica.
- Calor Específico ($C$): El pico del calor específico es el indicador más claro. Cuanto mayor es $N$, más alto y estrecho es el pico. En redes pequeñas, el pico es bajo y ancho, lo que dificulta la localización exacta de $T_c$.

4. <u>Reducción del Ruido Estadístico:</u> Las magnitudes termodinámicas calculadas son promedios. 
- En la red de $N=32$, el cambio de unos pocos espines tiene un impacto porcentual grande en la energía total, produciendo ruido en las gráficas.
- En $N=68$, las fluctuaciones locales se compensan mejor, dando lugar a curvas mucho más suaves y resultados más reproducibles.

## Animaciones GIF
Además de las capturas mostradas, se ha desarrollado una variante del código capaz de generar animaciones en tiempo real y exportar la evolución de las tres temperaturas de interés de forma simultánea en formato GIF para una evaluación dinámica más cómoda.

### Animación para dimensiones de red N = 32

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML, display # <--- NECESARIO PARA JUPYTER

# ==========================================
# PARÁMETROS DEL SISTEMA
# ==========================================
N = 32
T_values = [1.0, 2.27, 4.45]
update_interval = 50
FRAMES_TOTALES = 50  # <--- NOTA: He bajado los frames a 50 para que pruebes rápido.
                     # Luego súbelo a 200 cuando veas que funciona.

# (Las funciones inicializar_red y paso_monte_carlo son las mismas, las omito para abreviar pero déjalas en tu celda)
def inicializar_red(N, ordenada=False):
    return np.ones((N, N)) if ordenada else np.random.choice([-1, 1], size=(N, N))

def paso_monte_carlo(red, T):
    for _ in range(N * N):
        n, m = np.random.randint(0, N), np.random.randint(0, N)
        esp_actual = red[n, m]
        suma_vecinos = red[(n-1)%N, m] + red[(n+1)%N, m] + red[n, (m-1)%N] + red[n, (m+1)%N]
        delta_E = 2 * esp_actual * suma_vecinos
        if delta_E <= 0 or np.random.rand() < np.exp(-delta_E / T):
            red[n, m] = -esp_actual
    return red

# ==========================================
# CONFIGURACIÓN DE LA VISUALIZACIÓN
# ==========================================
reds = [inicializar_red(N, ordenada=False) for _ in T_values]
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
imagenes = []

for i, ax in enumerate(axs):
    ax.set_title(f"T = {T_values[i]}")
    ax.axis('off')
    im = ax.imshow(reds[i], cmap='coolwarm', interpolation='nearest', vmin=-1, vmax=1)
    imagenes.append(im)

def actualizar(frame):
    for i in range(3):
        reds[i] = paso_monte_carlo(reds[i], T_values[i])
        imagenes[i].set_data(reds[i])
    return imagenes

# CREAR LA ANIMACIÓN
anim = animation.FuncAnimation(fig, actualizar, frames=FRAMES_TOTALES, interval=update_interval, blit=True)

# CIERRA LA FIGURA para que no aparezca una imagen estática fea abajo
plt.close() 

# MOSTRAR EN JUPYTER
print("Generando animación en la celda...")
display(HTML(anim.to_jshtml())) # <--- Esto crea un reproductor interactivo en el notebook
```
